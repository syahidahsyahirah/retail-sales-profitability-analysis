```python
import matplotlib.pyplot as plt

# Chart 1: Profit margin by category
category_data = {
    "Technology": 17.40,
    "Office Supplies": 17.04,
    "Furniture": 2.49
}

colors = ["#4472C4" if v > 10 else "#C00000" for v in category_data.values()]

plt.figure(figsize=(6,4))
plt.bar(category_data.keys(), category_data.values(), color=colors)
plt.title("Profit Margin by Category")
plt.ylabel("Profit Margin (%)")
plt.axhline(0, color="black", linewidth=0.8)
plt.tight_layout()
plt.savefig("category_margin.png", dpi=150)
plt.show()
```

```python
# Chart 2: Furniture sub-category profit margin
subcat_data = {
    "Tables": -8.56,
    "Bookcases": -3.02,
    "Chairs": 8.10,
    "Furnishings": 14.24
}

colors2 = ["#C00000" if v < 0 else "#4472C4" for v in subcat_data.values()]

plt.figure(figsize=(6,4))
plt.bar(subcat_data.keys(), subcat_data.values(), color=colors2)
plt.title("Furniture: Profit Margin by Sub-Category")
plt.ylabel("Profit Margin (%)")
plt.axhline(0, color="black", linewidth=0.8)
plt.tight_layout()
plt.savefig("furniture_subcategory_margin.png", dpi=150)
plt.show()
```
